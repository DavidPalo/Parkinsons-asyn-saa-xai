```python
import pandas as pd
```

```python
df_24 = pd.read_csv("data/df_asociacion_24.csv")
df_150 = pd.read_csv("data/df_asociacion_150.csv")

df_mutacion1 = pd.read_csv("data/releases_2023_v4release_1027_amp_pd_participant_apoe_mutations.csv")
df_mutacion2 = pd.read_csv("data/releases_2023_v4release_1027_amp_pd_participant_mutations.csv")


df_mutacion1
```

Vamos a elegir en ambos dataframes unicamente los datos procedentes de PPMI

```python
df_mutacion1 = df_mutacion1[df_mutacion1["participant_id"].str.contains("PP-")]
df_mutacion2 = df_mutacion2[df_mutacion2["participant_id"].str.contains("PP-")]

df_mutacion = df_mutacion2

df_mutacion = df_mutacion.rename(columns={"participant_id": "PATNO"})
df_24["PATNO"] = df_24["PATNO"].astype(str)
df_mutacion["PATNO"] = df_mutacion["PATNO"].astype(str)
df_mutacion["PATNO"] = df_mutacion["PATNO"].str.replace("PP-", "", regex=False)

df_mutacion
```

```python
df_150["PATNO"] = df_150["PATNO"].astype(str)
```

Vamos a unir el .CSV que contiene la información con las mutaciones al .CSV que ya tenemos con el resto de los datos clínicos y de SAA

```python
df_24_completo = df_24.merge(
    df_mutacion,
    on=["PATNO"],
    how="left"
)

df_150_completo = df_150.merge(
    df_mutacion,
    on=["PATNO"],
    how="left"
)

df_24_completo = df_24_completo.sort_values(by="PATNO")
df_150_completo = df_150_completo.sort_values(by="PATNO")

bins = [0, 50, 60, 70, 120]
labels = ["<50", "50-59", "60-69", "≥70"]

df_24_completo["age_group"] = pd.cut(df_24_completo["age_at_baseline"], bins=bins, labels=labels)
df_150_completo["age_group"] = pd.cut(df_150_completo["age_at_baseline"], bins=bins, labels=labels)

col = df_24_completo.pop("age_group")
df_24_completo.insert(7, "age_group", col)
col = df_150_completo.pop("age_group")
df_150_completo.insert(7, "age_group", col)

df_24_completo.to_csv("df_24_mutaciones.csv")
df_150_completo.to_csv("df_150_mutaciones.csv")

df_150_completo

```
