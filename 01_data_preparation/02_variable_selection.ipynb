```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
df_full= pd.read_csv("dataframe_SAA_clinico_V0.csv")
df_clin = pd.read_csv("dataframe_clinicos_BL.csv")

df_150 = df_full[df_full["SAAMethod"] == "Amprion-Alpha-synuclein-SAA"].copy()
df_150 = df_150[df_150["SAA_Status"] != "Inconclusive"] #Aqui he quitado todos los resultados de la variable respuesta que aparecen como inconclusivos ya que no nos aportan ningún tipo de información.
df_150 = df_150[df_150["SAA_Type"] != "Undetermined"] #Aquí he eliminado todos aquellos en los que el SAA_Type salía como undetermined debido a que cuando sale undetermined se repite.
df_150 = df_150[df_150["CLINICAL_EVENT"] == "BL"]

df_24 = df_full[df_full["SAAMethod"] == "Amprion-24h alpha-synuclein-SAA"].copy()
df_24 = df_24[df_24["SAA_Status"] != "Inconclusive"]
df_24 = df_24[df_24["SAA_Type"] != "Undetermined"]
df_24 = df_24[df_24["CLINICAL_EVENT"] == "BL"]

df_clin = df_clin[df_clin["CLINICAL_EVENT"] == "BL"]
```

```python
id_col = "PATNO"     # cambia si tu ID se llama distinto
visit_col = "CLINICAL_EVENT"  # cambia si tu visita se llama distinto
bl_value = "BL"      # cambia si baseline es otro código

pos_sex_150 = df_150.columns.get_loc("SEX")
saa_cols_150 = list(df_150.columns[pos_sex_150+1:])
df_150_saa = df_150[["PATNO"] + saa_cols_150].copy()

pos_sex_24 = df_24.columns.get_loc("SEX")
saa_cols_24 = list(df_24.columns[pos_sex_24+1:])
df_24_saa = df_24[["PATNO"] + saa_cols_24].copy()

```

```python
df_clin_150 = df_clin.merge(
    df_150_saa,
    on=["PATNO"],
    how= "left"
)
df_clin_150 = df_clin_150.dropna(subset=["SAAMethod"])

df_clin_24 = df_clin.merge(
    df_24_saa,
    on=["PATNO"],
    how="left"
)
df_clin_24 = df_clin_24.dropna(subset=["SAAMethod"])



df_clin_24
```

##### Para poder trabajar en R unicamente con la variable respuesta y los datos clínicos, vamos a borrar todas las variables posteriores a la variable respuesta y todas aquellas variables de diagnosis que no tienen valores en la baseline

```python
df_onlyclin_150 = df_clin_150.loc[:, :'SAA_Status']

df_onlyclin_150
```

```python
df_onlyclin_150 = df_onlyclin_150.loc[:, ~df_onlyclin_150.columns.str.contains('_diagnosis', case=False)]
df_onlyclin_150 = df_onlyclin_150.loc[:, ~df_onlyclin_150.columns.str.contains('medication', case=False)]

df_onlyclin_150 = df_onlyclin_150.drop(columns=["surgery_for_parkinson_disease", "on_levodopa", "on_dopamine_agonist", "TYPE", "SAAMethod"])
df_onlyclin_150
```

```python
df_onlyclin_24 = df_clin_24.loc[:, :'SAA_Status']
df_onlyclin_24 = df_onlyclin_24.loc[:, ~df_onlyclin_24.columns.str.contains('_diagnosis', case=False)]
df_onlyclin_24 = df_onlyclin_24.loc[:, ~df_onlyclin_24.columns.str.contains('medication', case=False)]

df_onlyclin_24 = df_onlyclin_24.drop(columns=["surgery_for_parkinson_disease", "on_levodopa", "on_dopamine_agonist", "TYPE", "SAAMethod"])
df_onlyclin_24
```
