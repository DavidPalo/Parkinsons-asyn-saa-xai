```python
import pandas as pd
```

```python
df_150 = pd.read_csv("df_150_tabla.csv")
df_24 = pd.read_csv("df_24_tabla.csv")
```

```python


df_150 = df_150[df_150["study_arm"] != "SWEDD"]
df_24  = df_24[df_24["study_arm"] != "SWEDD"]

df_24[df_24["CLINICAL_EVENT"] != "BL"]
```

```python
bins = [0, 50, 60, 70, 80, float("inf")]
labels = ["<50", "50-59", "60-69", "70-79", ">=80"]
df_150["age_group"] = pd.cut(df_150["age_at_baseline"], bins=bins, labels=labels, right=False)
df_24["age_group"] = pd.cut(df_24["age_at_baseline"], bins=bins, labels=labels, right=False)
```

```python
df_150 = df_150.dropna(subset=["CLINICAL_EVENT"])
df_24 = df_24.dropna(subset=["CLINICAL_EVENT"])
df_24
```

```python
df_24.columns
```

```python
bl24 = df_24.pop("CLINICAL_EVENT")
bl150 = df_150.pop("CLINICAL_EVENT")
df_24.insert(2, "CLINICAL_EVENT", bl24)
df_150.insert(2, "CLINICAL_EVENT", bl150)

df_150
```

```python
df_150
```

```python

df_junto = pd.concat([df_150, df_24], ignore_index=True)

df_junto
```

```python
df_junto = df_junto.dropna(subset=["CLINICAL_EVENT"])

df_junto = df_junto.copy()
df_junto["CLINICAL_EVENT"] = df_junto["CLINICAL_EVENT"].replace("BL#2", "BL")
```
