```python
import pandas as pd
```

Cargamos los df correspondientes para 24 y 150 horas y eliminamos las columnas que no nos aporten información necesaria para el analisis

```python
df_24 = pd.read_csv("data/df_24_mutaciones.csv")
df_150 = pd.read_csv("data/df_150_mutaciones.csv")
df_24 = df_24.drop(df_24.columns[[0,1,2,4,5,6,12,14,15]], axis=1)
df_150 = df_150.drop(df_150.columns[[0,1,2,4,5,6,12,14,15]], axis=1)

#df_24 = df_24.dropna()
#df_150 = df_150.dropna()
```

```python
df_24
```

Ahora seleccionamos del dataset que contiene MDS-UPDRS-III score solo aquellos individuos correspondientes a ppmi y nos quedamos solo con la BL de cada paciente

```python
df_MDS = pd.read_csv("data/Dataset MDS-UPDRS-III score.csv")

df_MDS = df_MDS[df_MDS["participant_id"].str.startswith("PP-")]
df_MDS = df_MDS[df_MDS["visit_name"].str.startswith("SC")] # Selecciono solo la visita BL (en este df se llama SC)

df_MDS = df_MDS.rename(columns={"participant_id": "PATNO"})

df_MDS
```

```python
#Nos quedamos con la ultima columna del df que es lo unico que necesitamos

df_MDS = df_MDS[["PATNO", "visit_name", "mds_updrs_part_iii_summary_score"]]

df_MDS
```

Hacemos lo mismo para el resto de valores necesarios para la tabla

```python
df_HY = pd.read_csv("data/Dataset MDS-UPDRS-III score.csv")

df_HY = df_HY[df_HY["participant_id"].str.startswith("PP-")]
df_HY = df_HY[df_HY["visit_name"].str.startswith("SC")] # Selecciono solo la visita BL (en este df se llama SC)

df_HY = df_HY.rename(columns={"participant_id": "PATNO"})
df_HY = df_HY[["PATNO", "visit_name", "code_upd2hy_hoehn_and_yahr_stage"]]


df_HY
```

```python
df_moca = pd.read_csv("data/df_MoCa_Score.csv")

df_moca = df_moca[df_moca["participant_id"].str.startswith("PP-")]
df_moca = df_moca[df_moca["visit_name"].str.startswith("SC")] # Selecciono solo la visita BL (en este df se llama SC)

df_moca = df_moca.rename(columns={"participant_id": "PATNO"})

#Nos quedamos con la ultima columna del df que es lo unico que necesitamos

df_moca = df_moca[["PATNO", "visit_name", "moca_total_score"]]

df_moca
```

Hacemos merge entre los 3 nuevos datasets para crear uno solo que unir posteriormente a nuestros datasets de 24 y 150 horas con lo que podremos crear la tabla

```python
df_merge = df_MDS.merge(
    df_HY,
    on=["PATNO", "visit_name"],
    how="left"
)
```

```python
df_merge = df_merge.merge(
    df_moca,
    on=["PATNO", "visit_name"],
    how="outer"
)
```

```python
#Como hay varios pacientes en los que tenemos dos baselines, nos vamos a quedar con la segunda en caso de que estén repetidos

df_merge = (
    df_merge.assign(
        prioridad=df_merge["visit_name"].map({"SC#2": 2, "SC": 1}).fillna(0)
    )
    .sort_values(["PATNO", "prioridad"], ascending=[True, False])
    .drop_duplicates(subset="PATNO", keep="first")
    .drop(columns="prioridad")
)

```

```python
df_merge = df_merge.rename(columns={"visit_name": "CLINICAL_EVENT"})
df_merge["CLINICAL_EVENT"] = df_merge["CLINICAL_EVENT"].str.replace("SC", "BL")
df_merge["PATNO"] = df_merge["PATNO"].str.replace("PP-", "").astype(int)

df_merge
```

ahora vamos a hacer merge con nuestros df de 24 y 150 horas

```python
df_24_tabla = df_24.merge(
    df_merge,
    on = "PATNO",
    how = "left"
)

df_150_tabla = df_150.merge(
    df_merge,
    on = "PATNO",
    how = "left"
)

```

```python
df_150_tabla.to_csv("df_150_tabla.csv")
df_24_tabla.to_csv("df_24_tabla.csv")
```
