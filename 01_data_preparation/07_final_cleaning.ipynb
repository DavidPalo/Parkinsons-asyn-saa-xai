```python
import pandas as pd
```

```python
df_24 = pd.read_csv("df_24_tabla_completo.csv")
df_150 = pd.read_csv("df_150_tabla_completo.csv")

#Eliminamos todas las filas que tengan valores nulos
df_24 = df_24.dropna()
df_150 = df_150.dropna()
```

```python
#Eliminamos la columna Unnamed: 0 que se crea como indice adicional y que no nos es de ninguna utilidad

df_24 = df_24.drop(columns=["Unnamed: 0"])
df_150 = df_150.drop(columns=["Unnamed: 0"])
```

```python
#Guardamos los dataframes corregidos

df_150.to_csv("df_150_tabla.csv")
df_24.to_csv("df_24_tabla.csv")
```
