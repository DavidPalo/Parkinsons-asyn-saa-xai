vamos a proceder a imputar los missing values con 0 para las 3 vsariables de score en aquellos pacientes que no pertenezcan a algun grupo de la enfermedad de Parkinson. Y vamos a eliminar los participantes del estudio de aquellos en los que si

```python
import pandas as pd

import os
os.chdir(r'C:\Users\david\Desktop\master\TFM\CSV')
```

```python


# Cargar los dataframes
df_24 = pd.read_csv('df_24_completo_ordenado.csv')
df_150 = pd.read_csv('df_150_completo_ordenado.csv')

score_cols = ['mds_updrs_part_iii_summary_score', 'code_upd2hy_hoehn_and_yahr_stage', 'moca_total_score']

# Imputar con 0 los NaN de Healthy Control
for df in [df_24, df_150]:
    mask_hc = df['study_arm'] == 'Healthy Control'
    df.loc[mask_hc, score_cols] = df.loc[mask_hc, score_cols].fillna(0)

# Eliminar filas con NaN en scores del resto de grupos
filas_antes_24  = len(df_24)
filas_antes_150 = len(df_150)

df_24  = df_24.dropna(subset=score_cols)
df_150 = df_150.dropna(subset=score_cols)


# Verificación
print(f"df_24:  {filas_antes_24}  → {len(df_24)}  filas ({filas_antes_24  - len(df_24)}  eliminadas)")
print(f"df_150: {filas_antes_150} → {len(df_150)} filas ({filas_antes_150 - len(df_150)} eliminadas)")
print("\nNaN restantes en df_24:")
print(df_24[score_cols].isna().sum())
print("\nNaN restantes en df_150:")
print(df_150[score_cols].isna().sum())
print("\nstudy_arm df_24:")
print(df_24['study_arm'].value_counts())
print("\nstudy_arm df_150:")
print(df_150['study_arm'].value_counts())
```

```python
df_150.to_csv("df_150_score_imputado.csv")
df_24.to_csv("df_24_score_imputado.csv")
```

```python
# Añadir columna protocolo
df_24['protocolo']  = '24h'
df_150['protocolo'] = '150h'

# Concatenar
df_combined = pd.concat([df_24, df_150], ignore_index=True)

# Verificación
print(f"Filas df_24:      {len(df_24)}")
print(f"Filas df_150:     {len(df_150)}")
print(f"Filas combinado:  {len(df_combined)}")
print(f"\nDistribución protocolo:")
print(df_combined['protocolo'].value_counts())
```
